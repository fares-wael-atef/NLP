```python
"""
DSAI 353 Assignment 2: Enhanced RAG System for ArXiv Dataset
This notebook implements a streamlined Retrieval-Augmented Generation (RAG) system 
for the ArXiv Abstract & Title Dataset with clear documentation and analysis.

Technical specifications:
- Embedding model: sentence-transformers/all-mpnet-base-v2 (different from all-MiniLM-L6-v2)
- Generative model: google/flan-t5-large (different from mistralai/Mistral-7B-Instruct-v0.2)
"""

# Install required packages
!pip install -q transformers sentence-transformers rouge-score datasets torch numpy matplotlib pandas faiss-cpu

import os
import zipfile
import json
import time
from typing import List, Dict, Any, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from sentence_transformers import SentenceTransformer
import faiss
from rouge_score import rouge_scorer
from tqdm.notebook import tqdm

# Set random seed for reproducibility
np.random.seed(42)
torch.manual_seed(42)

# Part 1: Data Ingestion & Cleaning
def download_and_extract_dataset(url: str = "https://zenodo.org/records/3496527/files/gcunhase%2FArXivAbsTitleDataset-v1.0.zip?download=1", 
                             zip_path: str = "arxiv_abs_title.zip", 
                             extract_dir: str = "arxiv_dataset") -> str:
    """
    Download and extract the ArXiv dataset.
    
    Args:
        url: URL to download the dataset from
        zip_path: Path to save the downloaded file
        extract_dir: Directory to extract the dataset to
        
    Returns:
        Path to the extracted dataset JSON file
    """
    print(f"Downloading dataset from {url}...")
    !wget -q -O {zip_path} "{url}"
    
    print(f"Extracting dataset to {extract_dir}...")
    os.makedirs(extract_dir, exist_ok=True)
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    
    # Look for the dataset file
    possible_paths = [
        os.path.join(extract_dir, "ArXivAbsTitleDataset", "data", "arxiv_abs_title.json"),
        os.path.join(extract_dir, "data", "arxiv_abs_title.json"),
        os.path.join(extract_dir, "arxiv_abs_title.json")
    ]
    
    for path in possible_paths:
        if os.path.exists(path):
            print(f"Found dataset at: {path}")
            return path
    
    raise FileNotFoundError("Could not find the dataset file after extraction")

def load_dataset(file_path: str, sample_size: int = 1000) -> pd.DataFrame:
    """
    Load and clean the ArXiv dataset.
    
    Args:
        file_path: Path to the dataset JSON file
        sample_size: Number of samples to use (for faster processing)
        
    Returns:
        DataFrame containing the cleaned dataset
    """
    print(f"Loading dataset from {file_path}...")
    
    # Load the JSON data
    with open(file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    # Convert to DataFrame
    df = pd.DataFrame(data)
    
    # Clean the dataset
    # Handle encoding issues
    for col in ['title', 'abstract']:
        df[col] = df[col].apply(lambda x: x.encode('utf-8', 'ignore').decode('utf-8') if isinstance(x, str) else str(x))
    
    # Remove any rows with missing title or abstract
    df = df.dropna(subset=['title', 'abstract'])
    
    # Sample the dataset for faster processing
    if len(df) > sample_size:
        df = df.sample(sample_size, random_state=42)
    
    print(f"Dataset loaded and cleaned. Shape: {df.shape}")
    return df

# Part 2: Chunking Strategy
def chunk_documents(df: pd.DataFrame) -> List[Dict[str, Any]]:
    """
    Chunk documents using a document-level strategy.
    Each document (title + abstract) becomes one chunk.
    
    Args:
        df: DataFrame containing the documents
        
    Returns:
        List of chunk dictionaries
    """
    print("Chunking documents...")
    chunks = []
    
    for _, row in tqdm(df.iterrows(), total=len(df), desc="Creating chunks"):
        doc_id = row['id']
        title = row['title']
        abstract = row['abstract']
        categories = row['categories']
        
        # Combine title and abstract
        content = f"Title: {title}\nAbstract: {abstract}"
        
        chunks.append({
            'id': doc_id,
            'title': title,
            'content': content,
            'categories': categories
        })
    
    print(f"Created {len(chunks)} chunks")
    return chunks

# Part 3: Vectorization
def embed_chunks(chunks: List[Dict[str, Any]], model_name: str = 'sentence-transformers/all-mpnet-base-v2') -> Tuple[np.ndarray, SentenceTransformer]:
    """
    Embed chunks using the specified embedding model.
    
    Args:
        chunks: List of chunk dictionaries
        model_name: Name of the embedding model to use
        
    Returns:
        Tuple containing the chunk embeddings and the embedding model
    """
    print(f"Loading embedding model: {model_name}")
    embedding_model = SentenceTransformer(model_name)
    
    # Extract texts to embed
    texts = [chunk['content'] for chunk in chunks]
    
    # Measure time and memory usage
    start_time = time.time()
    
    # Get initial memory usage
    initial_memory = torch.cuda.memory_allocated() if torch.cuda.is_available() else 0
    
    # Embed chunks in batches
    batch_size = 32
    embeddings = []
    
    for i in tqdm(range(0, len(texts), batch_size), desc="Computing embeddings"):
        batch_texts = texts[i:i + batch_size]
        batch_embeddings = embedding_model.encode(batch_texts, show_progress_bar=False)
        embeddings.append(batch_embeddings)
    
    embeddings = np.vstack(embeddings)
    
    # Get final memory usage
    final_memory = torch.cuda.memory_allocated() if torch.cuda.is_available() else 0
    memory_used = (final_memory - initial_memory) / (1024 ** 2)  # Convert to MB
    
    end_time = time.time()
    elapsed_time = end_time - start_time
    
    print(f"Embeddings shape: {embeddings.shape}")
    print(f"Embedding time: {elapsed_time:.2f} seconds")
    if torch.cuda.is_available():
        print(f"Memory used: {memory_used:.2f} MB")
    
    return embeddings, embedding_model

# Part 4: Retrieval Module
def build_faiss_index(embeddings: np.ndarray) -> faiss.IndexFlatIP:
    """
    Build a FAISS index for the embeddings using inner product (cosine similarity).
    
    Args:
        embeddings: Chunk embeddings
        
    Returns:
        FAISS index
    """
    print("Building FAISS index...")
    
    # Normalize embeddings for cosine similarity
    faiss.normalize_L2(embeddings)
    
    # Build index
    embedding_dim = embeddings.shape[1]
    index = faiss.IndexFlatIP(embedding_dim)
    index.add(embeddings)
    
    print(f"Built FAISS index with {index.ntotal} vectors of dimension {embedding_dim}")
    return index

def retrieve_chunks(query: str, index: faiss.IndexFlatIP, embedding_model: SentenceTransformer, 
                   chunks: List[Dict[str, Any]], k: int = 5) -> List[Dict[str, Any]]:
    """
    Retrieve the top-k most relevant chunks for a query.
    
    Args:
        query: Query string
        index: FAISS index
        embedding_model: Embedding model
        chunks: List of chunk dictionaries
        k: Number of chunks to retrieve
        
    Returns:
        List of the top-k most relevant chunks with similarity scores
    """
    # Embed query
    query_embedding = embedding_model.encode([query])[0]
    
    # Normalize query embedding for cosine similarity
    query_embedding = query_embedding / np.linalg.norm(query_embedding)
    query_embedding = query_embedding.reshape(1, -1).astype(np.float32)
    
    # Search index
    scores, indices = index.search(query_embedding, k=min(k, len(chunks)))
    
    # Return top-k chunks with scores
    results = []
    for i, idx in enumerate(indices[0]):
        if idx < len(chunks):  # Ensure index is valid
            results.append({
                **chunks[idx],
                'similarity_score': float(scores[0][i])
            })
    
    return results

def create_test_queries(df: pd.DataFrame, num_queries: int = 10) -> List[Dict[str, Any]]:
    """
    Create test queries for retrieval evaluation.
    
    Args:
        df: DataFrame containing the documents
        num_queries: Number of test queries to create
        
    Returns:
        List of dictionaries, each containing a query and relevant chunk IDs
    """
    print(f"Creating {num_queries} test queries for evaluation...")
    
    # Get unique categories
    categories = df['categories'].str.split().explode().unique()
    selected_categories = np.random.choice(categories, min(5, len(categories)), replace=False)
    
    test_queries = []
    for category in selected_categories:
        # Find papers in this category
        category_papers = df[df['categories'].str.contains(category)].sample(min(2, len(df[df['categories'].str.contains(category)])))
        
        for _, paper in category_papers.iterrows():
            # Create a query from the title
            query = f"What are the main findings in research on {paper['title']}?"
            
            # The paper ID is considered relevant
            relevant_ids = [paper['id']]
            
            test_queries.append({
                'query': query,
                'relevant_ids': relevant_ids
            })
            
            if len(test_queries) >= num_queries:
                break
                
        if len(test_queries) >= num_queries:
            break
    
    return test_queries

def evaluate_retrieval(test_queries: List[Dict[str, Any]], index: faiss.IndexFlatIP, embedding_model: SentenceTransformer, 
                       chunks: List[Dict[str, Any]], max_k: int = 10) -> Dict[int, float]:
    """
    Evaluate the retrieval module using Top-k recall.
    
    Args:
        test_queries: List of dictionaries, each containing a query and relevant chunk IDs
        index: FAISS index
        embedding_model: Embedding model
        chunks: List of chunk dictionaries
        max_k: Maximum k value for Top-k recall
        
    Returns:
        Dictionary mapping k values to corresponding recall values
    """
    print("Evaluating retrieval performance...")
    
    recall_values = {k: [] for k in range(1, max_k + 1)}
    
    for query_info in tqdm(test_queries, desc="Testing queries"):
        query = query_info['query']
        relevant_ids = query_info['relevant_ids']
        
        # Get results for max_k
        results = retrieve_chunks(query, index, embedding_model, chunks, k=max_k)
        retrieved_ids = [result['id'] for result in results]
        
        # Calculate recall for each k
        for k in range(1, max_k + 1):
            retrieved_at_k = retrieved_ids[:k]
            relevant_retrieved = set(retrieved_at_k) & set(relevant_ids)
            recall = len(relevant_retrieved) / len(relevant_ids) if relevant_ids else 0
            recall_values[k].append(recall)
    
    # Average recall values across all queries
    avg_recall = {k: np.mean(values) for k, values in recall_values.items()}
    
    return avg_recall

def plot_recall_vs_k(recall_values: Dict[int, float]) -> None:
    """
    Plot Top-k recall vs. k.
    
    Args:
        recall_values: Dictionary mapping k values to corresponding recall values
    """
    k_values = list(recall_values.keys())
    avg_recall = list(recall_values.values())
    
    plt.figure(figsize=(10, 6))
    plt.plot(k_values, avg_recall, marker='o')
    plt.xlabel('k')
    plt.ylabel('Average Recall@k')
    plt.title('Top-k Recall vs. k')
    plt.grid(True)
    plt.xticks(k_values)
    plt.ylim(0, 1)
    plt.show()

# Part 5: Prompt Construction & Generation
def highlight_context(query: str, retrieved_chunks: List[Dict[str, Any]]) -> List[Dict[str, Any]]:
    """
    Highlight query terms in the retrieved chunks for better visualization.
    
    Args:
        query: Query string
        retrieved_chunks: List of retrieved chunk dictionaries
        
    Returns:
        List of dictionaries with highlighted content
    """
    # Extract query terms (excluding common words)
    stop_words = {'what', 'are', 'the', 'in', 'on', 'of', 'and', 'to', 'a', 'an', 'for', 'is', 'how'}
    query_terms = set(word.lower() for word in query.split() if word.lower() not in stop_words)
    
    highlighted_chunks = []
    for chunk in retrieved_chunks:
        content = chunk['content']
        
        # Split content into words and highlight query terms
        words = content.split()
        highlighted_words = []
        
        for word in words:
            word_clean = word.lower().strip(".,!?;:()[]{}\"'")
            if word_clean in query_terms:
                highlighted_words.append(f"**{word}**")
            else:
                highlighted_words.append(word)
        
        highlighted_content = ' '.join(highlighted_words)
        
        highlighted_chunks.append({
            **chunk,
            'highlighted_content': highlighted_content
        })
    
    return highlighted_chunks

def generate_answer(query: str, retrieved_chunks: List[Dict[str, Any]], model_name: str = 'google/flan-t5-large') -> str:
    """
    Generate an answer to the query using the retrieved chunks.
    
    Args:
        query: Query string
        retrieved_chunks: List of retrieved chunk dictionaries
        model_name: Name of the generative model to use
        
    Returns:
        Generated answer
    """
    print(f"Loading generation model: {model_name}")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
    
    # Construct context from retrieved chunks
    context_parts = []
    for i, chunk in enumerate(retrieved_chunks):
        context_parts.append(f"Document {i+1} (Score: {chunk['similarity_score']:.4f}):\n{chunk['content']}")
    
    context = "\n\n".join(context_parts)
    
    # Construct prompt
    prompt = f"""Context information:
{context}

Question: {query}

Answer the question based on the context information. If the answer cannot be determined from the context, say so."""
    
    # Generate answer
    inputs = tokenizer(prompt, return_tensors="pt", max_length=1024, truncation=True)
    outputs = model.generate(
        **inputs,
        max_length=512,
        num_beams=4,
        temperature=0.7,
        top_k=50,
        top_p=0.9,
    )
    answer = tokenizer.decode(outputs[0], skip_special_tokens=True)
    
    return answer

# Part 6: Evaluation & Reflection
def evaluate_answer_rouge(generated_answer: str, reference_answer: str) -> Dict[str, float]:
    """
    Evaluate the generated answer using ROUGE scores.
    
    Args:
        generated_answer: Generated answer
        reference_answer: Reference answer
        
    Returns:
        Dictionary containing ROUGE scores
    """
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
    scores = scorer.score(reference_answer, generated_answer)
    
    return {
        'rouge1': scores['rouge1'].fmeasure,
        'rouge2': scores['rouge2'].fmeasure,
        'rougeL': scores['rougeL'].fmeasure
    }

def run_rag_pipeline():
    """
    Run the complete RAG pipeline.
    """
    print("Starting RAG pipeline...")
    
    # 1. Data Ingestion & Cleaning
    try:
        dataset_path = download_and_extract_dataset()
        df = load_dataset(dataset_path)
    except Exception as e:
        print(f"Error in data ingestion: {e}")
        # Create a fallback dataset
        print("Creating fallback dataset for testing...")
        fallback_data = []
        for i in range(100):
            fallback_data.append({
                "id": f"fallback_{i}",
                "title": f"Deep Learning Architecture {i}",
                "abstract": f"This paper introduces a novel deep learning architecture for computer vision. We propose a new approach to feature extraction that improves performance on benchmark tasks.",
                "categories": "cs.AI cs.CV"
            })
        df = pd.DataFrame(fallback_data)
    
    # 2. Chunking Strategy
    chunks = chunk_documents(df)
    
    print("\nChunking Strategy Justification:")
    print("- Document-level chunking was chosen because:")
    print("  * Each title + abstract combination forms a coherent, self-contained unit")
    print("  * The typical abstract length is manageable for embedding models")
    print("  * This maintains the natural context and semantic unity of research papers")
    print("  * Simplifies retrieval by preserving document-level information")
    
    # 3. Vectorization
    embeddings, embedding_model = embed_chunks(chunks)
    
    # 4. Retrieval Module
    index = build_faiss_index(embeddings)
    
    # Create test queries and evaluate retrieval
    test_queries = create_test_queries(df)
    recall_values = evaluate_retrieval(test_queries, index, embedding_model, chunks)
    
    # Plot recall vs. k
    plot_recall_vs_k(recall_values)
    
    # 5. Prompt Construction & Generation
    print("\nDemonstrating Q&A examples with context highlighting:")
    example_queries = [
        "What are the latest advancements in computer vision?",
        "How can reinforcement learning be applied to robotics?",
        "What are the challenges in natural language understanding?",
        "How do transformers work in deep learning?",
        "What are the applications of graph neural networks?"
    ]
    
    for i, query in enumerate(example_queries):
        print(f"\nExample Q&A {i+1}:")
        print(f"Query: {query}")
        
        # Retrieve chunks
        retrieved_chunks = retrieve_chunks(query, index, embedding_model, chunks, k=3)
        
        # Highlight query terms in the retrieved chunks
        highlighted_chunks = highlight_context(query, retrieved_chunks)
        
        # Display highlighted chunks
        print("\nRetrieved chunks:")
        for j, chunk in enumerate(highlighted_chunks):
            print(f"\nChunk {j+1} (Score: {chunk['similarity_score']:.4f}):")
            print(chunk['highlighted_content'])
        
        # Generate answer
        answer = generate_answer(query, retrieved_chunks)
        print(f"\nGenerated Answer: {answer}")
    
    # 6. Evaluation & Reflection
    print("\nEvaluation & Reflection:")
    
    # Create test cases with reference answers
    test_cases = [
        {
            'query': "What are the main applications of deep learning in computer vision?",
            'reference_answer': "Deep learning is applied in computer vision for tasks such as image classification, object detection, image segmentation, and facial recognition."
        },
        {
            'query': "How do transformer models work?",
            'reference_answer': "Transformer models work using self-attention mechanisms that allow them to process all input tokens in parallel while capturing dependencies between them, replacing traditional recurrent architectures."
        },
        {
            'query': "What are the challenges in reinforcement learning?",
            'reference_answer': "Challenges in reinforcement learning include sample inefficiency, exploration-exploitation trade-off, non-stationarity, and the difficulty of designing appropriate reward functions."
        }
    ]
    
    # Evaluate test cases
    print("\nQuantitative Evaluation:")
    all_scores = {'rouge1': [], 'rouge2': [], 'rougeL': []}
    
    for i, case in enumerate(test_cases):
        query = case['query']
        reference_answer = case['reference_answer']
        
        # Retrieve chunks
        retrieved_chunks = retrieve_chunks(query, index, embedding_model, chunks, k=3)
        
        # Generate answer
        generated_answer = generate_answer(query, retrieved_chunks)
        
        # Evaluate answer
        scores = evaluate_answer_rouge(generated_answer, reference_answer)
        
        print(f"\nTest Case {i+1}:")
        print(f"Query: {query}")
        print(f"Generated: {generated_answer}")
        print(f"Reference: {reference_answer}")
        print(f"ROUGE Scores: R1={scores['rouge1']:.4f}, R2={scores['rouge2']:.4f}, RL={scores['rougeL']:.4f}")
        
        # Add scores to aggregates
        for metric, score in scores.items():
            all_scores[metric].append(score)
    
    # Calculate average scores
    avg_scores = {metric: np.mean(scores) for metric, scores in all_scores.items()}
    print(f"\nAverage ROUGE Scores: R1={avg_scores['rouge1']:.4f}, R2={avg_scores['rouge2']:.4f}, RL={avg_scores['rougeL']:.4f}")
    
    # Qualitative reflection
    print("\nQualitative Reflection:")
    print("1. Strengths of our RAG system:")
    print("   - Successfully retrieves relevant papers based on semantic similarity")
    print("   - Generates coherent answers that utilize context from retrieved papers")
    print("   - Document-level chunking preserves the coherence of research concepts")
    print("   - MPNet embeddings capture semantic relationships effectively")
    
    print("\n2. Limitations and failure cases:")
    print("   - Vocabulary mismatch between queries and papers affects retrieval")
    print("   - Limited capacity to synthesize information across multiple papers")
    print("   - May generate plausible-sounding but incorrect answers if retrieved context is insufficient")
    print("   - Struggles with highly technical or domain-specific queries")
    
    print("\n3. Potential improvements:")
    print("   - Implement hybrid retrieval (combining dense and sparse methods)")
    print("   - Add query expansion to address vocabulary mismatch")
    print("   - Introduce a re-ranking step after initial retrieval")
    print("   - Fine-tune the language model on scientific literature")
    print("   - Implement a strategy to detect and handle unanswerable questions")

if __name__ == "__main__":
    run_rag_pipeline()
```

```python
def build_faiss_index(embeddings: np.ndarray) -> faiss.IndexFlatIP:
    """
    Build a FAISS index for the embeddings using inner product (cosine similarity).
    
    Args:
        embeddings: Chunk embeddings
        
    Returns:
        FAISS index
    """
    print("Building FAISS index...")
    
    # Normalize embeddings for cosine similarity
    faiss.normalize_L2(embeddings)
    
    # Build index
    embedding_dim = embeddings.shape[1]
    index = faiss.IndexFlatIP(embedding_dim)
    index.add(embeddings)
    
    print(f"Built FAISS index with {index.ntotal} vectors of dimension {embedding_dim}")
    return index
```

## Retrieve Chunks

```python
def retrieve_chunks(query: str, index: faiss.IndexFlatIP, embedding_model: SentenceTransformer, 
                   chunks: List[Dict[str, Any]], k: int = 5) -> List[Dict[str, Any]]:
    """
    Retrieve the top-k most relevant chunks for a query.
    
    Args:
        query: Query string
        index: FAISS index
        embedding_model: Embedding model
        chunks: List of chunk dictionaries
        k: Number of chunks to retrieve
        
    Returns:
        List of the top-k most relevant chunks with similarity scores
    """
    # Embed query
    query_embedding = embedding_model.encode([query])[0]
    
    # Normalize query embedding for cosine similarity
    query_embedding = query_embedding / np.linalg.norm(query_embedding)
    query_embedding = query_embedding.reshape(1, -1).astype(np.float32)
    
    # Search index
    scores, indices = index.search(query_embedding, k=min(k, len(chunks)))
    
    # Return top-k chunks with scores
    results = []
    for i, idx in enumerate(indices[0]):
        if idx < len(chunks):  # Ensure index is valid
            results.append({
                **chunks[idx],
                'similarity_score': float(scores[0][i])
            })
    
    return results
```

## Create Test Queries

```python
def create_test_queries(df: pd.DataFrame, num_queries: int = 10) -> List[Dict[str, Any]]:
    """
    Create test queries for retrieval evaluation.
    
    Args:
        df: DataFrame containing the documents
        num_queries: Number of test queries to create
        
    Returns:
        List of dictionaries, each containing a query and relevant chunk IDs
    """
    print(f"Creating {num_queries} test queries for evaluation...")
    
    # Get unique categories
    categories = df['categories'].str.split().explode().unique()
    selected_categories = np.random.choice(categories, min(5, len(categories)), replace=False)
    
    test_queries = []
    for category in selected_categories:
        # Find papers in this category
        category_papers = df[df['categories'].str.contains(category)].sample(min(2, len(df[df['categories'].str.contains(category)])))
        
        for _, paper in category_papers.iterrows():
            # Create a query from the title
            query = f"What are the main findings in research on {paper['title']}?"
            
            # The paper ID is considered relevant
            relevant_ids = [paper['id']]
            
            test_queries.append({
                'query': query,
                'relevant_ids': relevant_ids
            })
            
            if len(test_queries) >= num_queries:
                break
                
        if len(test_queries) >= num_queries:
            break
    
    return test_queries
```

## Evaluate Retrieval

```python
def evaluate_retrieval(test_queries: List[Dict[str, Any]], index: faiss.IndexFlatIP, embedding_model: SentenceTransformer, 
                       chunks: List[Dict[str, Any]], max_k: int = 10) -> Dict[int, float]:
    """
    Evaluate the retrieval module using Top-k recall.
    
    Args:
        test_queries: List of dictionaries, each containing a query and relevant chunk IDs
        index: FAISS index
        embedding_model: Embedding model
        chunks: List of chunk dictionaries
        max_k: Maximum k value for Top-k recall
        
    Returns:
        Dictionary mapping k values to corresponding recall values
    """
    print("Evaluating retrieval performance...")
    
    recall_values = {k: [] for k in range(1, max_k + 1)}
    
    for query_info in tqdm(test_queries, desc="Testing queries"):
        query = query_info['query']
        relevant_ids = query_info['relevant_ids']
        
        # Get results for max_k
        results = retrieve_chunks(query, index, embedding_model, chunks, k=max_k)
        retrieved_ids = [result['id'] for result in results]
        
        # Calculate recall for each k
        for k in range(1, max_k + 1):
            retrieved_at_k = retrieved_ids[:k]
            relevant_retrieved = set(retrieved_at_k) & set(relevant_ids)
            recall = len(relevant_retrieved) / len(relevant_ids) if relevant_ids else 0
            recall_values[k].append(recall)
    
    # Average recall values across all queries
    avg_recall = {k: np.mean(values) for k, values in recall_values.items()}
    
    return avg_recall
```

## Plot Recall vs. k

```python
def plot_recall_vs_k(recall_values: Dict[int, float]) -> None:
    """
    Plot Top-k recall vs. k.
    
    Args:
        recall_values: Dictionary mapping k values to corresponding recall values
    """
    k_values = list(recall_values.keys())
    avg_recall = list(recall_values.values())
    
    plt.figure(figsize=(10, 6))
    plt.plot(k_values, avg_recall, marker='o')
    plt.xlabel('k')
    plt.ylabel('Average Recall@k')
    plt.title('Top-k Recall vs. k')
    plt.grid(True)
    plt.xticks(k_values)
    plt.ylim(0, 1)
    plt.show()
```

## Highlight Context

```python
def highlight_context(query: str, retrieved_chunks: List[Dict[str, Any]]) -> List[Dict[str, Any]]:
    """
    Highlight query terms in the retrieved chunks for better visualization.
    
    Args:
        query: Query string
        retrieved_chunks: List of retrieved chunk dictionaries
        
    Returns:
        List of dictionaries with highlighted content
    """
    # Extract query terms (excluding common words)
    stop_words = {'what', 'are', 'the', 'in', 'on', 'of', 'and', 'to', 'a', 'an', 'for', 'is', 'how'}
    query_terms = set(word.lower() for word in query.split() if word.lower() not in stop_words)
    
    highlighted_chunks = []
    for chunk in retrieved_chunks:
        content = chunk['content']
        
        # Split content into words and highlight query terms
        words = content.split()
        highlighted_words = []
        
        for word in words:
            word_clean = word.lower().strip(".,!?;:()[]{}\"'")
            if word_clean in query_terms:
                highlighted_words.append(f"**{word}**")
            else:
                highlighted_words.append(word)
        
        highlighted_content = ' '.join(highlighted_words)
        
        highlighted_chunks.append({
            **chunk,
            'highlighted_content': highlighted_content
        })
    
    return highlighted_chunks
```

## Generate Answer

```python
def generate_answer(query: str, retrieved_chunks: List[Dict[str, Any]], model_name: str = 'google/flan-t5-large') -> str:
    """
    Generate an answer to the query using the retrieved chunks.
    
    Args:
        query: Query string
        retrieved_chunks: List of retrieved chunk dictionaries
        model_name: Name of the generative model to use
        
    Returns:
        Generated answer
    """
    print(f"Loading generation model: {model_name}")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
    
    # Construct context from retrieved chunks
    context_parts = []
    for i, chunk in enumerate(retrieved_chunks):
        context_parts.append(f"Document {i+1} (Score: {chunk['similarity_score']:.4f}):\n{chunk['content']}")
    
    context = "\n\n".join(context_parts)
    
    # Construct prompt
    prompt = f"""Context information:
{context}

Question: {query}

Answer the question based on the context information. If the answer cannot be determined from the context, say so."""
    
    # Generate answer
    inputs = tokenizer(prompt, return_tensors="pt", max_length=1024, truncation=True)
    outputs = model.generate(
        **inputs,
        max_length=512,
        num_beams=4,
        temperature=0.7,
        top_k=50,
        top_p=0.9,
    )
    answer = tokenizer.decode(outputs[0], skip_special_tokens=True)
    
    return answer
```

## Evaluate Answer with ROUGE

```python
def evaluate_answer_rouge(generated_answer: str, reference_answer: str) -> Dict[str, float]:
    """
    Evaluate the generated answer using ROUGE scores.
    
    Args:
        generated_answer: Generated answer
        reference_answer: Reference answer
        
    Returns:
        Dictionary containing ROUGE scores
    """
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
    scores = scorer.score(reference_answer, generated_answer)
    
    return {
        'rouge1': scores['rouge1'].fmeasure,
        'rouge2': scores['rouge2'].fmeasure,
        'rougeL': scores['rougeL'].fmeasure
    }
```

## Run RAG Pipeline

```python
def run_rag_pipeline():
    """
    Run the complete RAG pipeline.
    """
    print("Starting RAG pipeline...")
    
    # 1. Data Ingestion & Cleaning
    try:
        dataset_path = download_and_extract_dataset()
        df = load_dataset(dataset_path)
    except Exception as e:
        print(f"Error in data ingestion: {e}")
        # Create a fallback dataset
        print("Creating fallback dataset for testing...")
        fallback_data = []
        for i in range(100):
            fallback_data.append({
                "id": f"fallback_{i}",
                "title": f"Deep Learning Architecture {i}",
                "abstract": f"This paper introduces a novel deep learning architecture for computer vision. We propose a new approach to feature extraction that improves performance on benchmark tasks.",
                "categories": "cs.AI cs.CV"
            })
        df = pd.DataFrame(fallback_data)
    
    # 2. Chunking Strategy
    chunks = chunk_documents(df)
    
    print("\nChunking Strategy Justification:")
    print("- Document-level chunking was chosen because:")
    print("  * Each title + abstract combination forms a coherent, self-contained unit")
    print("  * The typical abstract length is manageable for embedding models")
    print("  * This maintains the natural context and semantic unity of research papers")
    print("  * Simplifies retrieval by preserving document-level information")
    
    # 3. Vectorization
    embeddings, embedding_model = embed_chunks(chunks)
    
    # 4. Retrieval Module
    index = build_faiss_index(embeddings)
    
    # Create test queries and evaluate retrieval
    test_queries = create_test_queries(df)
    recall_values = evaluate_retrieval(test_queries, index, embedding_model, chunks)
    
    # Plot recall vs. k
    plot_recall_vs_k(recall_values)
    
    # 5. Prompt Construction & Generation
    print("\nDemonstrating Q&A examples with context highlighting:")
    example_queries = [
        "What are the latest advancements in computer vision?",
        "How can reinforcement learning be applied to robotics?",
        "What are the challenges in natural language understanding?",
        "How do transformers work in deep learning?",
        "What are the applications of graph neural networks?"
    ]
    
    for i, query in enumerate(example_queries):
        print(f"\nExample Q&A {i+1}:")
        print(f"Query: {query}")
        
        # Retrieve chunks
        retrieved_chunks = retrieve_chunks(query, index, embedding_model, chunks, k=3)
        
        # Highlight query terms in the retrieved chunks
        highlighted_chunks = highlight_context(query, retrieved_chunks)
        
        # Display highlighted chunks
        print("\nRetrieved chunks:")
        for j, chunk in enumerate(highlighted_chunks):
            print(f"\nChunk {j+1} (Score: {chunk['similarity_score']:.4f}):")
            print(chunk['highlighted_content'])
        
        # Generate answer
        answer = generate_answer(query, retrieved_chunks)
        print(f"\nGenerated Answer: {answer}")
    
    # 6. Evaluation & Reflection
    print("\nEvaluation & Reflection:")
    
    # Create test cases with reference answers
    test_cases = [
        {
            'query': "What are the main applications of deep learning in computer vision?",
            'reference_answer': "Deep learning is applied in computer vision for tasks such as image classification, object detection, image segmentation, and facial recognition."
        },
        {
            'query': "How do transformer models work?",
            'reference_answer': "Transformer models work using self-attention mechanisms that allow them to process all input tokens in parallel while capturing dependencies between them, replacing traditional recurrent architectures."
        },
        {
            'query': "What are the challenges in reinforcement learning?",
            'reference_answer': "Challenges in reinforcement learning include sample inefficiency, exploration-exploitation trade-off, non-stationarity, and the difficulty of designing appropriate reward functions."
        }
    ]
    
    # Evaluate test cases
    print("\nQuantitative Evaluation:")
    all_scores = {'rouge1': [], 'rouge2': [], 'rougeL': []}
    
    for i, case in enumerate(test_cases):
        query = case['query']
        reference_answer = case['reference_answer']
        
        # Retrieve chunks
        retrieved_chunks = retrieve_chunks(query, index, embedding_model, chunks, k=3)
        
        # Generate answer
        generated_answer = generate_answer(query, retrieved_chunks)
        
        # Evaluate answer
        scores = evaluate_answer_rouge(generated_answer, reference_answer)
        
        print(f"\nTest Case {i+1}:")
        print(f"Query: {query}")
        print(f"Generated: {generated_answer}")
        print(f"Reference: {reference_answer}")
        print(f"ROUGE Scores: R1={scores['rouge1']:.4f}, R2={scores['rouge2']:.4f}, RL={scores['rougeL']:.4f}")
        
        # Add scores to aggregates
        for metric, score in scores.items():
            all_scores[metric].append(score)
    
    # Calculate average scores
    avg_scores = {metric: np.mean(scores) for metric, scores in all_scores.items()}
    print(f"\nAverage ROUGE Scores: R1={avg_scores['rouge1']:.4f}, R2={avg_scores['rouge2']:.4f}, RL={avg_scores['rougeL']:.4f}")
    
    # Qualitative reflection
    print("\nQualitative Reflection:")
    print("1. Strengths of our RAG system:")
    print("   - Successfully retrieves relevant papers based on semantic similarity")
    print("   - Generates coherent answers that utilize context from retrieved papers")
    print("   - Document-level chunking preserves the coherence of research concepts")
    print("   - MPNet embeddings capture semantic relationships effectively")
    
    print("\n2. Limitations and failure cases:")
    print("   - Vocabulary mismatch between queries and papers affects retrieval")
    print("   - Limited capacity to synthesize information across multiple papers")
    print("   - May generate plausible-sounding but incorrect answers if retrieved context is insufficient")
    print("   - Struggles with highly technical or domain-specific queries")
    
    print("\n3. Potential improvements:")
    print("   - Implement hybrid retrieval (combining dense and sparse methods)")
    print("   - Add query expansion to address vocabulary mismatch")
    print("   - Introduce a re-ranking step after initial retrieval")
    print("   - Fine-tune the language model on scientific literature")
    print("   - Implement a strategy to detect and handle unanswerable questions")
```

## Main Execution

```python
if __name__ == "__main__":
    run_rag_pipeline()
```
